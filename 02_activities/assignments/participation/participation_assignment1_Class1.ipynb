## Making a basic figure with matplotlib

```python
# Load our libraries:
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
    ## Had to run `uv add scipy --active` and `uv add requests --active` to get this to work with our environment
```

```python
# Make some sample data so that we have something to plot:
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)

# Make our basic scatterplot using plt.subplots
fig, ax = plt.subplots(figsize=(5, 3)) #specify figure size; 5 is width and 3 is height
    ## creates two objects (fig & ax; Figure and axes) ->  1. figure object (overall container for entire plot; all axis, titles, labels, etc) & 2. axes object (actual plotting area you plot your data in)
ax.scatter(x,y) #plots a scatter plot on the axes object `ax`, our figure
    ## Could do `plt.scatter(x,y)` without creating fig, ax but this way gives you more control (size, multiple subplots, formatting...)
```

```python
#We can also make a bar plot:
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
#Or a line plot:
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
#And a histogram:
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y,edgecolor='black') #only need y value, shows frequency of y values
    ##Note that here, `hist` expects just the one data (not x,y but y); can also give optional parameter of `bins=`
```

## Plot Labels

```python
#We can also plot labels - Like adding axis labels and titles

#Let’s add to the line plot we made above:
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout() #automatically adjusts spacing between subplots, axes, titles, ticks and labels so that nothing overlaps
```

```python
# Modifying labels and titles
## We can modify our labels and titles’ appearance by using font dictionaries (‘fontdict’)
## First, make a dictionary with the properties we want for our labels
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}

# Then we’ll add fontdict arguments to our previous lineplot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
#Moving labels and titles
## We can move our lables with the loc argument
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1,loc='left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

## Customizing plot appearance

```python
# Modifying data points

#Remake our basic scatterplot, but this time add color and marker arguments (Note: American spelling)
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y,
marker='*', # will determine shape of the point
color = "indigo") # color of the point
fig.show() #fig.show is intended to display figures outside of interactive environments (like normal Python script run from the terminal) so this ran in VSC will result in warning

# For VSC, use plt.show()
```

```python
# We can also adjust the style and width of the line connecting our points
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',color = 'indigo',
linestyle = '--', # Style of dashes
linewidth = 2) # width of the connections
plt.show()
```

```python
# We can modify colour using named colors using: https://matplotlib.org/stable/gallery/color/named_colors.html#sphx-glr-gallerycolor-named-colors-py
# Or we can use hex codes
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x,y,marker='*', 
color = '#AEC6CF' #color being the new addition, to light blue
,linestyle = '--', linewidth = 2)
plt.show()
```

```python
# Let’s customize our plot further
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x,y,marker='*',
markersize = 12, # size of points
color = '#AEC6CF',linestyle = '--',linewidth = 2,
markeredgecolor = '#fa9359', #changes color of the edge of data points
markerfacecolor = '#000000') #changes color of the marker
plt.show()
```

## Gridlines

```python
#We can add grid lines to our axis object using .grid()
ax.grid(axis = 'y')

#Cam modify the appearance (colour, width, linestyle) of our grid lines using what we learned before:
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',color = "indigo") 
ax.grid(axis = 'y')
ax.grid(axis = 'y'
, color = "blue"
, linewidth = 2, linestyle = '-.')
plt.show()
```

```python
# Import the packages
import pandas as pd
import holoviews as hv
from holoviews import opts, dim
from bokeh.sampledata.les_mis import data
```

```python
# Should the diagram be plotted with 'bokeh' or 'matplotlib'?
hv.extension('bokeh')
# How large should be the diagram?
hv.output(size=300)

# Data set
nodes = hv.Dataset(pd.DataFrame(data['nodes']), 'index')
links = pd.DataFrame(data['links'])

# Chord diagram
chord = hv.Chord((links, nodes)).select(value=(5, None))
chord.opts(
    opts.Chord(cmap='Category20', edge_cmap='Category20', edge_color=dim('source').str(), 
               labels='name', node_color=dim('index').str()))

# Not needed in a jupyter notebook
# It shows the diagram when run in another IDE or from a python script:
from bokeh.plotting import show
show(hv.render(chord))
```
