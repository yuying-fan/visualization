## Legends

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613) 
x = np.arange(50) 
y1 = np.random.randint(0, 100,50) 
y2 = np.random.randint(0, 100,50)
```

```python
## Make a line plot with both of our y variables on the same axes
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.plot(x,y1) 
ax.plot(x,y2) 
fig.show()
```

```python
## To add a legend:
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.plot(x,y1, label = "Person 1" ) 
ax.plot(x,y2, label = "Person 2" ) 
ax.legend(loc='lower right') 
fig.show()
```

```python
## To modify the legend:
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right', 
	frameon = True, #add frame around the legend 
	fontsize = 12, #change font size 
	ncol = 2, #specify number of columns 
	shadow = True ) #shadow on margins of legend
fig.show()
```

```python
## Moving legend outside of the plot area:
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.plot(x,y1, label = "Person 1") 
ax.plot(x,y2, label = "Person 2") 
ax.legend(loc='upper left', 
	bbox_to_anchor =(1, 1)) 
fig.show()
```

## Text and Annotations

```python
# Annotations with the text() function:
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x,y1, label = "Person 1") 
ax.scatter(x,y2, label = "Person 2") 
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!") 
fig.show()
```

```python
# Modifying text annotations:
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x,y1, label = "Person 1") 
ax.scatter(x,y2, label = "Person 2") 
ax.legend(loc='lower right',
	prop={"family":"serif"
	,"size":5}) ##need dictionary to specify font type and size
ax.text(10, 95, "This value is important!", 
	ha='center', ##horizontal alignment of text 
	color = 'red', ##modify text colour 
	size = 20) ##modify font size 
fig.show()
```

```python
# Positioning text annotations:
fig, ax = plt.subplots() 
ax.axis([0, 10, 0, 10])

ax.text(1, 5, ". Data:(1, 5)", transform=ax.transData ) 
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes ) 
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure )
```

```python
#Annotating with arrows:
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1") 
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
	arrowprops=dict(facecolor='black')) 
fig.show()
```

```python
# Modifying arrows:
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1") 
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.annotate('This is important!', 
	xy=(10, 95), xytext=(20, 94), 
	arrowprops = dict(arrowstyle = "wedge", color = "hotpink")) 
fig.show()
```

## Axis Labels

```python
## Removing tick marks and labels:
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x,y1, label = "Person 1") 
ax.scatter(x,y2, label = "Person 2") 
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
## Limiting tick marks:
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
## Tick mark intervals:
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x,y1, label = "Person 1") 
ax.scatter(x,y2, label = "Person 2") 
ax.legend(loc='lower right') 
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
fig,(ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3)) 
ax1.scatter(x1,y1) #scatter plot for left, ax1
ax2.bar(x2,y2) #bar plot for right, ax2
fig.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7, 4)) 
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
```

### Multiple visuals on one axes

```python
# first make our sample data 
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"]) 
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50]) 

# define our figure and axes (just one this time) 
fig, ax = plt.subplots(figsize=(7, 3)) 

# now call both bar and plot elements to the same axes (ax) 
ax.bar(x, y1, color = "indigo") 
ax.plot(x, y2, color = "red")
```

### Add error information

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3)) 
ax.plot(x, y2, color = "red")
```

```python
fig, ax = plt.subplots(figsize=(7, 3)) 
ax.errorbar(x, #our x values 
y2, #our y values 
color="blue",
yerr = y2_sd, 
fmt = "none")
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.errorbar(x, y2, color="red", yerr = y2_sd,
fmt = "-", ecolor='blue') #for line. o = dot/scatter
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red") 
ax.errorbar(x, y2, color="blue", yerr = y2_sd, fmt = "none")
```

### Customizing errorbar appearance

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, y2, yerr = y2_sd, fmt = "none", ecolor= "indigo", elinewidth= 4, capsize = 6, capthick= 4 )
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, y2, yerr = y2_sd, fmt = "none", ecolor = "indigo", elinewidth = 4, capsize = 6, capthick = 4, errorevery= 2)
```

```python
#Sometimes wikipedia will block getting image so this line is needed:
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"
headers = {'User-Agent': 'Mozilla/5.0',
           "Accept": 'image/*,*/*;q=0.8',
           "Referer": "https://en.wikipedia.org"}

response = requests.get(url, headers=headers)
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

### Wordclouds and Venn diagrams

```python
from wordcloud import WordCloud 
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv", on_bad_lines='skip') 
df
```

```python
# join all our text from each row from our quote column into a string 
text = " ".join(each for each in df.quote) 

# generate our wordcloud image 
wordcloud = WordCloud(background_color="white", colormap = 'inferno').generate(text) 
# use matplotlib syntax to put our image in a figure 
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud, #'imshow' for adding pictures to matplotlib axes 
interpolation='bilinear') #helps smooth our image 
ax.axis("off")
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
```

```python
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
```

```python
diagram = venn2_unweighted([A, B], set_labels = ('Fruits', 'Pies'), set_colors=("blue", "red"), alpha=0.5) 
plt.show()
```

```python
diagram = venn2_unweighted([A, B], set_labels = ('Fruits', 'Pies'), set_colors=("blue", "red"), alpha=0.5) 
diagram.get_label_by_id("10").set_text("\n".join(A - B))
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))
plt.show()
```
